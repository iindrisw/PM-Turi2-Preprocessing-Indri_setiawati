```python
import numpy as np
import pandas as pd

data = {
    'usia': [25, 27, np.nan, 35, 40, np.nan, 48],
    'pendapatan': [5000000, np.nan, 7000000, 10000000, np.nan, 12000000, 15000000],
    'pendidikan': ['SMA', 'S1', 'S2', 'S1', 'SMA', 'S2', 'S1'],
    'kota': ['Jakarta', 'Bandung', 'Jakarta', 'Surabaya', 'Bandung', 'Jakarta', 'Surabaya'],
    'membeli': ['Tidak', 'Ya', 'Tidak', 'Ya', 'Tidak', 'Ya', 'Ya']
}

df = pd.DataFrame(data)

df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
print(df)
```

```python
x = df.drop (columns=['membeli'])
y = df['membeli']
print(x)
print(y)
```

```python
x['pendidikan']=x['pendidikan'].map({'SMA':0, 'S1':1, 'S2':2})
x = pd.get_dummies(x,columns=['kota'], dtype=int)
y = y.map({'Tidak':0, 'Ya':1})
print(x)
print(y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x,y,test_size=0.3, random_state=42
)
print(x)
```

```python
from sklearn.preprocessing import StandardScaler
num = ['usia','pendapatan','pendidikan']
sc= StandardScaler()
x_train[num] = sc.fit_transform(x_train[num])
x_test[num] = sc.transform(x_test[num])
print(x_test)
print(x_train)
```

4. Jelaskan mengapa fit_transform pada latih tetapi hanya transform pada uji.

Jawaban: Alasan nya untuk mencegah kebocoran data dan menjaga pengujian model tetap objektif.
fit_transform pada data latih:
fit berguna untuk  parameter statistik dari data latih.transform bertugas mengubah/menyesuaikan skala data menggunakan parameter yang baru dipelajari tersebut.
transform pada data uji: Data uji hanya diskalakan menggunakan nilai rata-rata dan standar deviasi milik data latih

refleksi
1. mengapa urutan 'split dulu, baru scaling' penting?
Tujuan dari data uji adalah menjadi simulasi data riil yang belum pernah dilihat sama sekali oleh model. Jika data uji ikut dipakai untuk menghitung skala, hasil metrik performa akan menjadi palsu karena model secara tidak langsung sudah terbiasa dengan karakteristik sebaran data uji.
2. kapan label encoding dan kapan one hot encoding
label encoding cocok untuk data yang ordinal (urutan/mempunyai level) contohnya smp,sma,s1,s2
label one hot encoding cocok untuk data yang nominal(tanpa urutan) contohnya surabaya, cirebon
3. apa perbedaan normalisasi dan standarisasi
normalisasi berdasarkan min dan max (0-1), berguna bila butuh data jelas
standarisasi berdasarkan z-score, berguna jika tidak membatasi rentang
